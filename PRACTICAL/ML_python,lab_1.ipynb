1.Write a Python program to classify a student's CGPA into High, Mid, or Low categories using if-elif-else.

```python
cgpa = 8.2

if cgpa >= 8.0:
    category = "High"
elif cgpa >= 6.0:
    category = "Mid"
else:
    category = "Low"

print(f"CGPA: {cgpa}, Category: {category}")
```

2.Write a program to check whether missing values are present in a dataset using conditional statements.

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Score': [85, None, 92]
})

if df.isnull().values.any():
    print("Missing values are present in the dataset.")
else:
    print("No missing values found in the dataset.")
```

4.Write a program that uses if-elif-else to assign grades based on marks.

```python
marks = 85

if marks >= 90:
    grade = "A+"
elif marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 50:
    grade = "D"
else:
    grade = "F"

print(f"Marks: {marks}, Grade: {grade}")
```

2.Write a program to calculate the average of numbers stored in a list.

```python
numbers = [12, 45, 67, 23, 89, 34]
total_sum = 0

for num in numbers:
    total_sum += num

average = total_sum / len(numbers)
print(f"Average: {average}")
```

4.Write a program to find the maximum and minimum values in a list using loops.

```python
numbers = [45, 12, 89, 3, 67, 34]

max_val = numbers[0]
min_val = numbers[0]

for num in numbers[1:]:
    if num > max_val:
        max_val = num
    if num < min_val:
        min_val = num

print(f"Maximum value: {max_val}")
print(f"Minimum value: {min_val}")
```

5.Write a program to count the number of elements satisfying a given condition using a loop.

```python
numbers = [12, 7, 18, 5, 22, 9, 14]
count = 0

for num in numbers:
    if num % 2 == 0:
        count += 1

print(f"Count of even numbers: {count}")
```
